```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
Matplotlib Basic
```

```python
x = np.linspace(0, 5, 11)
x
y = x ** 4
y
```

```python
plt.plot(x, y)
plt.title("Hello") #Title of this plot
plt.xlabel("A axis")
plt.ylabel("Y axis")
```

```python
SubPlots in Matplotlib
```

```python
plt.subplot(2, 2, 1)
plt.plot(x,y)
plt.subplot(2, 2, 4)
plt.plot(x, y)
plt.subplot(2, 2, 3)
plt.plot(y, x)
plt.subplot(2, 2, 2)
plt.plot(x**2, x**3)
```
